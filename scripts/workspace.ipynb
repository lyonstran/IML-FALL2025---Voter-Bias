```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import ttest_ind
```

```python
pollster_df = pd.read_csv(r'C:\Users\Lyons\OneDrive\Desktop\IML-FALL2025---Voter-Bias\webscraped_data\pollsters_2014_2024.csv')
mi_cutoff_df = pd.read_csv(r'C:\Users\Lyons\OneDrive\Desktop\IML-FALL2025---Voter-Bias\voter_team\voter_team_cutoff50_weighted_michigan.csv')
```

```python
pollster_df = pollster_df.rename(columns = {'pollster_name' : 'voter'})
pollster_df['voter'] = pollster_df['voter'].str.lower()
pollster_df['voter'] = pollster_df['voter'].str.replace(' ', '-')
pollster_df
```

```python
df = mi_cutoff_df.merge(pollster_df[['voter', 'city', 'state', 'year']], on = 'voter', how = 'left')
df.year = df.year.astype('str').str.replace('.0', '')
df
```

```python
team_state_map = {
    'alabama' : 'AL', 
    'michigan' : 'MI', 
    'ohio-state' : 'OH', 
    'lsu' : 'LA', 
    'oklahoma' : 'OK'
}

bias_column = 'bias1' # add more columns? 
results = []

for team, team_state in team_state_map.items():
    temp = df.copy()
    temp['in_state'] = temp['state'].eq(team_state)

    summary = {
        "team": team,
        "team_state": team_state,
        "avg_bias_in_state": temp[temp["in_state"]][bias_column].mean(),
        "avg_bias_out_state": temp[temp["in_state"] == False][bias_column].mean(),
        "n_in_state": temp[temp["in_state"]].shape[0],
        "n_out_state": temp[temp["in_state"] == False].shape[0]
    }
    results.append(summary)

analysis_df = pd.DataFrame(results)
analysis_df
```
